```python
import pandas as pd
```

```python
df = pd.read_csv('../ball_positions.csv')
df.head(300)
```

```python
import numpy as np

# Tune these values
window_size = 10      # number of previous valid detections used as reference
max_jump = 200
# maximum allowed jump in pixels

df_new = df.copy()

recent_valid = []
outlier_rows = []

for i, row in df_new.iterrows():
    x, y = row["x_center"], row["y_center"]

    if pd.isna(x) or pd.isna(y):
        continue

    if len(recent_valid) >= window_size:
        ref = np.median(np.array(recent_valid[-window_size:]), axis=0)
        dist = np.hypot(x - ref[0], y - ref[1])

        if dist > max_jump:
            df_new.loc[i, ["x_center", "y_center"]] = np.nan
            outlier_rows.append(i)
            continue

    recent_valid.append([x, y])

print(f"Removed {len(outlier_rows)} false detections: {outlier_rows}")
df_new.head(300)
```
